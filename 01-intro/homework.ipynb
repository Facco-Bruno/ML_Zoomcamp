# Q6

```python
# 1. Mediana inicial da coluna horsepower
median_before = df["horsepower"].median()

# 2. Valor mais frequente (moda) da coluna horsepower
mode_value = df["horsepower"].mode()[0]

# 3. Preencher valores ausentes com a moda
df["horsepower"].fillna(mode_value, inplace=True)

# 4. Calcular a nova mediana
median_after = df["horsepower"].median()

print("Median before:", median_before)
print("Most frequent (mode):", mode_value)
print("Median after:", median_after)

# 5. Verificar se mudou
if median_after > median_before:
    print("Yes, it increased")
elif median_after < median_before:
    print("Yes, it decreased")
else:
    print("No")
```

# Q7

```python
# 1. Selecionar apenas carros da Ásia
asia_cars = df[df["origin"] == "Asia"]

# 2. Selecionar colunas vehicle_weight e model_year
asia_subset = asia_cars[["vehicle_weight", "model_year"]]

# 3. Selecionar os primeiros 7 valores
X = asia_subset.head(7).to_numpy()

# 4. Calcular XTX
XTX = X.T @ X

# 5. Inverter XTX
XTX_inv = np.linalg.inv(XTX)

# 6. Criar y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 7. Calcular w = (XTX_inv @ X.T) @ y
w = XTX_inv @ X.T @ y

# 8. Soma dos elementos
result = w.sum()

print("Resultado final:", result)
```
